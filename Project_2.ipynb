```python
def get_item_data():
    name = input("Enter item name: ")
    category = input("Enter category: ")
    price = input("Enter price: ")
    while float(price) < 0:
        price = input("Invalid price. Please enter the correct price: ")
    item = {"Name": name, "Category": category, "Price": price}
    return item



def print_item(item):
    print(item["Name"] + "  |  " + item["Category"] + "  |  " + "$" + item["Price"])


def calculate_total_with_tax(items):
    total = 0
    for item in items:
        total += float(item["Price"])
    tax = float(input("Enter tax rate in %: "))
    if tax == 0:
        return total
    while tax < 0 or tax > 90:
        tax = float(input("Invalid tax rate. Enter the correct tax rate in %: "))
    return total + total * tax / 100

def calculate_total(items):
    total = 0
    for item in items:
        total += float(item["Price"])
    return total



def calculate_average_price(items):
    return calculate_total(items) / len(items)


def find_most_expensive(items):
    idx = 0
    for i in range(1, len(items)):
        if float(items[i]["Price"]) > float(items[idx]["Price"]):
            idx = i
    return items[idx]


def find_cheapest(items):
    idx = 0
    for i in range(1, len(items)):
        if float(items[i]["Price"]) < float(items[idx]["Price"]):
            idx = i
    return items[idx]


def filter_by_category(items, category):
    list = []
    for item in items:
        if item["Category"].strip().lower() == category.strip().lower():
            list.append(item)
    return list


def main():
    num = int(input("How many items? "))
    while num < 1:
        num = int(input("Invalid number. Please enter the correct amount: "))
    items = []
    for i in range(num):
        print()
        items.append(get_item_data())
    print()
    print("All Items: ")
    for item in items:
        print_item(item)
    print()
    tax = float(input("Enter tax rate in %: "))
    total = calculate_total(items)
    if tax == 0:
        total = calculate_total(items)
    while tax < 0 or tax > 90:
        tax = float(input("Invalid tax rate. Enter the correct tax rate in %: "))
    total += total * tax / 100
    print("Receipt Summary: ")
    print("Total Cost: $" + str(total))
    print("Average Cost: $" + str(total / len(items)))
    print()
    print("Most Expensive Item: ")
    print_item(find_most_expensive(items))
    print()
    print("Cheapest Item: ")
    print_item(find_cheapest(items))
    print()
    category = input("Enter category to search: ")
    print()
    l = filter_by_category(items, category)
    if l == []:
        print("No items found in that category.")
    else:
        print("Items in category '" + category +"':")
        for i in l:
            print_item(i)

    
    




main()
```
